```python
from ultralytics import YOLO

model = YOLO('yolov26n.pt')  # Lightweight for faster training

# Train
model.train(data="Z:\Arjun k p\9 Facial Expressions you need\data.yaml", epochs=10, imgsz=640, batch=16, name='emotion_model', save=True)
```

```python
# Evaluate on validation set
metrics = model.val()

# Show all available metrics
print(metrics.box)  # precision, recall, mAP50, mAP50-95
```

```python
# Predict on test images
results = model.predict(source=r"Z:\Arjun k p\9 Facial Expressions you need\test\images", save=True, conf=0.25)
```
